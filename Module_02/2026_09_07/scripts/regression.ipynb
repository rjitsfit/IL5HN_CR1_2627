```python
import numpy as np
```

```python
class LinearRegression:
    def __init__ (self):
        self.intercept_ : float = 0.0
        self.slope_ : float = 0.0

    def fit (self, X: np.ndarray, y: np.ndarray) -> tuple[float, float]:
        X_flat = X.flatten ()
        y_flat = y.flatten ()

        X_mean = np.mean (X_flat)
        y_mean = np.mean (y_flat)


        numerator = np.sum ((X_flat - X_mean)*(y_flat - y_mean))
        denominator = np.sum ((X_flat - X_mean)**2)

        self.slope_ = numerator / denominator
        self.intercept_ = y_mean - (self.slope_ * X_mean)

        return self.intercept_, self.slope_

    def predict (self, Xi):
        return self.intercept_ + (self.slope_ * Xi)
```

```python
X_data = np.array ([160, 171, 182, 180, 154])
y_data = np.array ([72, 76, 77, 83, 76])
```

```python
model = LinearRegression ()
b0, b1 = model.fit (X_data, y_data)
```

```python
print (f'Intercept : {b0} and slope : {b1}')
```

```python
print(model.predict(176))
```

```python
def mean_squared_error (y_true, y_pred):
    y_true_arr = np.asarray (y_true).flatten()
    y_pred_arr = np.asarray (y_pred).flatten()
    return np.mean ((y_true_arr - y_pred_arr)**2)
```

```python
y_pred = model.predict (X_data)
```

```python
mse = mean_squared_error (y_data, y_pred)
print (f'Mean Squared Error is {mse}')
```

```python
def r2_score (y_true, y_pred):
    y_true_arr = np.asarray (y_true).flatten()
    y_pred_arr = np.asarray (y_pred).flatten()

    sse = np.sum ((y_true_arr - y_pred_arr)**2)
    sst = np.sum ((y_true_arr - np.mean(y_true_arr))**2)
    ssr = np.sum ((y_pred_arr - np.mean(y_pred_arr))**2)
    # return (1 - sse/sst)
    return ssr/sst
```

```python
result = r2_score(y_data, y_pred)
print (f'Goodness of Fit : {result}')
```
